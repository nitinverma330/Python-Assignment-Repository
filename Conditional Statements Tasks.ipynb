### Task 1: ATM Withdrawal System
**Conditions:**
1. Check whether PIN is correct (default `1234`).
2. If PIN is correct, check whether withdrawal amount is a multiple of 100.
3. If valid amount, check whether balance is sufficient.
4. If sufficient, deduct amount and display updated balance.
5. Otherwise, display appropriate error message (`Insufficient Balance`, `Invalid Amount`, `Invalid PIN`).

```python
def atm_withdrawal(entered_pin, withdrawal_amount, account_balance=25000):
    CORRECT_PIN = 1234
    
    print(f"--- ATM Transaction ---")
    print(f"Current Balance: Rs. {account_balance}")
    print(f"Requested Withdrawal: Rs. {withdrawal_amount}")
    
    if entered_pin == CORRECT_PIN:
        if withdrawal_amount % 100 == 0 and withdrawal_amount > 0:
            if withdrawal_amount <= account_balance:
                account_balance -= withdrawal_amount
                print(f"Transaction Successful! Please collect your cash.")
                print(f"Remaining Balance: Rs. {account_balance}")
            else:
                print("Transaction Failed: Insufficient Balance (Paisa Nhi Hai!)")
        else:
            print("Transaction Failed: Amount must be a positive multiple of 100.")
    else:
        print("Transaction Failed: Invalid PIN entered.")
    print("-" * 30)

# Test cases
atm_withdrawal(1234, 2000, 25000)   # Valid transaction
atm_withdrawal(1234, 350, 25000)    # Not multiple of 100
atm_withdrawal(1234, 30000, 25000)  # Insufficient balance
atm_withdrawal(9999, 1000, 25000)   # Wrong PIN
```

### Task 2: Electricity Bill Calculation
**Conditions:**
- 0 - 100 units: ₹5 / unit
- 101 - 200 units: ₹7 / unit
- 201 - 300 units: ₹10 / unit
- Above 300 units: ₹12 / unit
- **Additional:** If total bill > ₹3000, add a 15% surcharge.

```python
def calculate_electricity_bill(units):
    print(f"Units Consumed: {units}")
    
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = (100 * 5) + ((units - 100) * 7)
    elif units <= 300:
        bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)
    else:
        bill = (100 * 5) + (100 * 7) + (100 * 10) + ((units - 300) * 12)
        
    surcharge = 0.0
    if bill > 3000:
        surcharge = bill * 0.15
        total_bill = bill + surcharge
        print(f"Base Bill: Rs. {bill:.2f}")
        print(f"Surcharge (15% for bill > 3000): Rs. {surcharge:.2f}")
        print(f"Total Amount Payable: Rs. {total_bill:.2f}")
    else:
        total_bill = bill
        print(f"Total Amount Payable: Rs. {total_bill:.2f}")
    print("-" * 35)

# Test cases
calculate_electricity_bill(80)    # <= 100
calculate_electricity_bill(150)   # 101-200
calculate_electricity_bill(250)   # 201-300
calculate_electricity_bill(350)   # > 300 (surcharge applies)
```

### Task 3: Employee Bonus Eligibility System
**Conditions:**
- **Experience >= 5 years:**
  - If Salary >= ₹50,000:
    - Performance >= 90: 20% bonus
    - Performance >= 75: 10% bonus
    - Otherwise: 5% bonus
  - If Salary < ₹50,000:
    - Performance >= 90: 15% bonus
    - Otherwise: 8% bonus
- **Experience < 5 years:**
  - Performance >= 90: 10% bonus
  - Otherwise: 5% bonus

```python
def calculate_employee_bonus(name, experience_years, salary, performance_score):
    print(f"Employee: {name} | Exp: {experience_years} yrs | Salary: Rs.{salary} | Score: {performance_score}")
    
    if experience_years >= 5:
        if salary >= 50000:
            if performance_score >= 90:
                bonus_pct = 0.20
            elif performance_score >= 75:
                bonus_pct = 0.10
            else:
                bonus_pct = 0.05
        else:
            if performance_score >= 90:
                bonus_pct = 0.15
            else:
                bonus_pct = 0.08
    else:
        if performance_score >= 90:
            bonus_pct = 0.10
        else:
            bonus_pct = 0.05
            
    bonus_amount = salary * bonus_pct
    print(f"Bonus Percentage: {bonus_pct * 100:.0f}% | Bonus Amount: Rs. {bonus_amount:.2f}")
    print(f"Total Compensation: Rs. {salary + bonus_amount:.2f}")
    print("-" * 45)

# Test cases
calculate_employee_bonus("Rahul", 6, 60000, 92)
calculate_employee_bonus("Priya", 5, 45000, 80)
calculate_employee_bonus("Amit", 2, 40000, 95)
calculate_employee_bonus("Neha", 3, 35000, 70)
```

### Task 4: Student Result System (3 Subjects with Cutoff & Divisions)
**Conditions:**
- Takes marks of 3 subjects: Java, Python, DBMS (0 - 100).
- Every subject must have marks >= 40 to pass. If any subject < 40, student fails in that subject.
- If passed all subjects:
  - Percentage >= 75: Distinction / Honors
  - Percentage >= 60: First Division
  - Percentage >= 50: Second Division
  - Percentage >= 40: Third Division

```python
def student_result_system(java, python, dbms):
    print(f"Marks -> Java: {java}, Python: {python}, DBMS: {dbms}")
    
    if java < 40:
        print("Result: Failed in Java")
    elif python < 40:
        print("Result: Failed in Python")
    elif dbms < 40:
        print("Result: Failed in DBMS")
    else:
        total_marks = java + python + dbms
        percentage = total_marks / 3.0
        print(f"Total Marks: {total_marks} / 300 | Percentage: {percentage:.2f}%")
        
        if percentage >= 75:
            print("Division: First Division with Distinction")
        elif percentage >= 60:
            print("Division: First Division")
        elif percentage >= 50:
            print("Division: Second Division")
        else:
            print("Division: Third Division (Pass)")
    print("-" * 40)

# Test cases
student_result_system(85, 90, 88)
student_result_system(65, 72, 61)
student_result_system(52, 45, 54)
student_result_system(77, 34, 80)  # Failed in Python
```

### Task 5: Bank Loan Eligibility
**Conditions:**
- Age must be >= 21.
- Salary must be >= ₹25,000.
- If age and salary are valid:
  - Credit score >= 750: Loan Approved (Best interest rate)
  - Credit score >= 650: Loan Approved with Higher Interest Rate
  - Otherwise: Loan Rejected due to Low Credit Score
- If age < 21: Ineligible due to age
- If salary < ₹25,000: Ineligible due to income criteria

```python
def check_loan_eligibility(age, salary, credit_score):
    print(f"Applicant Info -> Age: {age}, Salary: Rs. {salary}, Credit Score: {credit_score}")
    
    if age >= 21:
        if salary >= 25000:
            if credit_score >= 750:
                print("Status: Loan Approved at Standard Interest Rate.")
            elif credit_score >= 650:
                print("Status: Loan Approved with Higher Interest Rate.")
            else:
                print("Status: Loan Rejected (Credit score must be at least 650).")
        else:
            print("Status: Loan Rejected (Minimum monthly salary of Rs. 25,000 required).")
    else:
        print("Status: Loan Rejected (Applicant must be at least 21 years old).")
    print("-" * 45)

# Test cases
check_loan_eligibility(28, 45000, 780)
check_loan_eligibility(24, 30000, 680)
check_loan_eligibility(30, 50000, 580)
check_loan_eligibility(19, 35000, 720)
```

### Task 6: E-Commerce Discount Engine
**Conditions:**
- Cart Amount >= ₹5000:
  - Premium Member: 20% discount
  - Regular Member: 10% discount
- Cart Amount >= ₹2000:
  - Premium Member: 10% discount
  - Regular Member: 5% discount
- Otherwise (Cart Amount < ₹2000): No discount

```python
def calculate_ecommerce_discount(cart_amount, is_premium_member):
    member_type = "Premium" if is_premium_member else "Regular"
    print(f"Cart Amount: Rs. {cart_amount} | Membership: {member_type}")
    
    if cart_amount >= 5000:
        if is_premium_member:
            discount_pct = 0.20
        else:
            discount_pct = 0.10
    elif cart_amount >= 2000:
        if is_premium_member:
            discount_pct = 0.10
        else:
            discount_pct = 0.05
    else:
        discount_pct = 0.00
        
    discount_amount = cart_amount * discount_pct
    final_amount = cart_amount - discount_amount
    print(f"Discount ({discount_pct*100:.0f}%): Rs. {discount_amount:.2f}")
    print(f"Final Payable Amount: Rs. {final_amount:.2f}")
    print("-" * 40)

# Test cases
calculate_ecommerce_discount(6500, True)
calculate_ecommerce_discount(6500, False)
calculate_ecommerce_discount(3000, True)
calculate_ecommerce_discount(1200, True)
```

### Task 7: Role-Based Login & Access Control
**Conditions:**
- Valid credentials verification (`username` and `password`).
- If valid, grant access based on `role`:
  - `admin`: Full Admin Dashboard Access
  - `teacher`: Teacher / Grading Portal
  - `student`: Student Learning Dashboard
  - any other: Invalid Role
- If password incorrect: "Invalid Password"
- If username incorrect: "Invalid Username"

```python
# User database mock
USERS_DB = {
    "admin_user": {"password": "admin@123", "role": "admin"},
    "john_doe": {"password": "pass456", "role": "student"},
    "prof_smith": {"password": "teach789", "role": "teacher"}
}

def login_system(username, password):
    print(f"Login Attempt -> User: '{username}'")
    
    if username in USERS_DB:
        user_record = USERS_DB[username]
        if password == user_record["password"]:
            role = user_record["role"].lower()
            if role == "admin":
                print("Access Granted: Welcome Admin! Redirecting to Full Control Dashboard.")
            elif role == "teacher":
                print("Access Granted: Welcome Teacher! Redirecting to Grading & Course Portal.")
            elif role == "student":
                print("Access Granted: Welcome Student! Redirecting to Student Dashboard.")
            else:
                print("Access Denied: Unrecognized Role.")
        else:
            print("Access Denied: Invalid Password.")
    else:
        print("Access Denied: Invalid Username (User not found).")
    print("-" * 45)

# Test cases
login_system("admin_user", "admin@123")
login_system("john_doe", "pass456")
login_system("john_doe", "wrongpass")
login_system("unknown_user", "12345")
```

### Task 8: Movie Ticket Booking System
**Conditions:**
- Check seat availability: if `available_seats <= 0`, display "House Full!".
- If seats are available, determine price based on age and seat tier:
  - Age < 12: Child ticket (50% off standard)
  - Age >= 60: Senior citizen ticket (30% off standard)
  - Regular adult ticket
  - Seat Tier: VIP adds ₹150 surcharge; Standard has no surcharge.

```python
def book_movie_ticket(age, seat_tier="Standard", available_seats=5):
    BASE_PRICE = 300
    print(f"Booking Request -> Age: {age}, Tier: {seat_tier}, Available Seats: {available_seats}")
    
    if available_seats <= 0:
        print("Booking Failed: House Full! No seats available.")
    else:
        # Age based discount
        if age < 12:
            price = BASE_PRICE * 0.50
            category = "Child (50% discount)"
        elif age >= 60:
            price = BASE_PRICE * 0.70
            category = "Senior Citizen (30% discount)"
        else:
            price = BASE_PRICE
            category = "Adult"
            
        # Seat tier surcharge
        if seat_tier.lower() == "vip":
            price += 150
            tier_desc = "VIP (+Rs. 150)"
        else:
            tier_desc = "Standard"
            
        available_seats -= 1
        print(f"Ticket Confirmed! Category: {category} | Tier: {tier_desc}")
        print(f"Total Ticket Price: Rs. {price:.2f} | Remaining Seats: {available_seats}")
    print("-" * 45)

# Test cases
book_movie_ticket(8, "Standard", 10)
book_movie_ticket(65, "VIP", 5)
book_movie_ticket(25, "VIP", 2)
book_movie_ticket(30, "Standard", 0)  # House full
```

### Task 9: Positive, Negative, or Zero Number Classifier

```python
def check_number(num):
    if num > 0:
        print(f"{num} is Positive.")
    elif num < 0:
        print(f"{num} is Negative.")
    else:
        print("The number is Zero.")

check_number(10)
check_number(-5)
check_number(0)
```

### Task 10: Age Voting Eligibility & Campus ID Verification

```python
def verify_voting_and_campus_entry(age, has_id_card):
    print(f"Age: {age}, Has ID Card: {has_id_card}")
    
    # Voting eligibility check
    if age >= 18:
        print("Voting Status: Eligible to Vote.")
    else:
        print("Voting Status: Not Eligible to Vote (Must be 18+).")
        
    # Campus entry check with nested if
    if age >= 18:
        if has_id_card:
            print("Campus Entry: Student Entry Allowed.")
        else:
            print("Campus Entry: Entry Denied (Valid Student ID Card Required).")
    else:
        print("Campus Entry: Age criteria not met for independent campus access.")
    print("-" * 40)

verify_voting_and_campus_entry(20, True)
verify_voting_and_campus_entry(22, False)
verify_voting_and_campus_entry(16, True)
```
