## Part A: Dictionary Fundamentals & Methods

```python
# 1. Creation
d1 = {}
d2 = dict()
print("d1:", d1, "| type:", type(d1))
print("d2:", d2, "| type:", type(d2))

# 2. Key uniqueness and value duplication
student = {"name1": "Mohit", "age": 23, "active": True, "name2": "Mohit"}
print("\nDuplicate values allowed:", student)

# Duplicate keys overwrite previous values
student_overwrite = {"name": "Mohit", "age": 23, "name": "Amit"}
print("Duplicate key overwritten:", student_overwrite)

# 3. Mutability & accessing keys
data = {"name": "Mohit", "age": 23, "Att": True}
data["name"] = "Priya"  # In-place update
data["city"] = "Bareilly"  # Adding new key
print("\nUpdated dictionary:", data)

print("Direct key access data['name']:", data["name"])
print("Safe access data.get('college'):", data.get("college"))  # Returns None (no KeyError)
print("Safe access with default:", data.get("college", "Invertis University"))
```

### Dictionary Inspection and Modification Methods

```python
data = {"name": "Mohit", "age": 23, "Att": True}

print("Length len(data):", len(data))
print("Keys data.keys():", list(data.keys()))
print("Values data.values():", list(data.values()))
print("Items data.items():", list(data.items()))

# pop() removes specific key and returns its value
age = data.pop("age")
print(f"\nRemoved 'age' ({age}):", data)

# popitem() removes the last inserted key-value pair
last_item = data.popitem()
print(f"Removed last item {last_item}:", data)

# clear() empties the dictionary
data.clear()
print("After clear():", data)
```

### Shallow Copy vs Reference Assignment

```python
original = {"name": "Mohit", "age": 23, "Att": True}

# Reference assignment: Both point to the same memory object
ref_data = original
ref_data["name"] = "Rohit"
print("Modifying ref_data also modified original:")
print("original:", original)
print("ref_data:", ref_data)

# Shallow copy: Independent dictionary object
original = {"name": "Mohit", "age": 23, "Att": True}
copied_data = original.copy()
copied_data["name"] = "Rohit"
print("\nUsing .copy():")
print("original:   ", original)     # Still 'Mohit'
print("copied_data:", copied_data)  # 'Rohit'
```

### Task 1: Student Pass / Fail Classifier
Given a dictionary of student marks, display whether each student passed or failed (Passing marks >= 40):

```python
students = {
    "Rahul": 75,
    "Amit": 32,
    "Priya": 88,
    "Neha": 39,
    "Rohit": 55
}

print("--- Task 1: Student Pass / Fail ---")
for name, marks in students.items():
    if marks >= 40:
        print(f"{name:6s} : {marks:2d} -> Pass")
    else:
        print(f"{name:6s} : {marks:2d} -> Fail")
```

### Task 2: Student Performance Grading Using Nested If
**Conditions:**
- `Marks >= 80`: Excellent
- `Marks >= 60`: Good
- `Marks >= 40`: Average
- `Marks < 40`: Fail

```python
students = {
    "Rahul": 85,
    "Amit": 35,
    "Priya": 72,
    "Neha": 48
}

print("--- Task 2: Student Performance ---")
for name, marks in students.items():
    if marks >= 40:
        if marks >= 80:
            category = "Excellent"
        elif marks >= 60:
            category = "Good"
        else:
            category = "Average"
    else:
        category = "Fail"
    print(f"{name:6s} : {marks:2d} -> {category}")
```

### Task 3: Product Stock Inventory Management
**Conditions:**
- `Stock == 0`: Out of Stock
- `Stock <= 5`: Low Stock
- `Stock > 5`: In Stock

```python
products = {
    "Laptop": 10,
    "Keyboard": 0,
    "Monitor": 2,
    "Printer": 8,
    "Mouse": 4
}

print("--- Task 3: Product Stock Status ---")
for product, stock in products.items():
    if stock == 0:
        status = "Out of Stock (Reorder Immediately!)"
    elif stock <= 5:
        status = f"Low Stock ({stock} units remaining)"
    else:
        status = f"In Stock ({stock} units)"
    print(f"{product:10s} : {status}")
```

### Task 4: Employee Salary & Department Role Classifier
**Conditions:**
- `Salary >= 50,000`:
  - If department is `"IT"`: Senior IT Employee
  - If department is `"HR"`: Senior HR Employee
  - Otherwise: Senior Employee
- `Salary < 50,000`: Junior Employee

```python
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit":  {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha":  {"salary": 45000, "department": "Finance"},
    "Vikas": {"salary": 55000, "department": "Marketing"}
}

print("--- Task 4: Employee Categories ---")
for name, data in employees.items():
    salary = data["salary"]
    dept = data["department"]
    
    if salary >= 50000:
        if dept == "IT":
            category = "Senior IT Employee"
        elif dept == "HR":
            category = "Senior HR Employee"
        else:
            category = f"Senior Employee ({dept})"
    else:
        category = f"Junior Employee ({dept})"
        
    print(f"{name:6s} (Rs.{salary:,}, {dept:9s}) -> {category}")
```

### Task 5: Check Even / Odd & Large Values
Check whether each dictionary value is even or odd, and flag if it is a large even number (> 20):

```python
data = {
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}

print("--- Task 5: Even / Odd Classification ---")
for key, val in data.items():
    if val % 2 == 0:
        if val > 20:
            classification = "Large Even Number"
        else:
            classification = "Even Number"
    else:
        classification = "Odd Number"
    print(f"Key '{key}' ({val:2d}) -> {classification}")
```

### Task 6: Shopping Cart Price Tier Classification
**Conditions:**
- `Price >= 50000`: Luxury Premium Product
- `Price >= 5000`: Premium Product
- `Price >= 1000`: Expensive Product
- `Price < 1000`: Affordable Product

```python
products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200,
    "Headphones": 3500
}

print("--- Task 6: Product Price Tier ---")
for item, price in products.items():
    if price >= 50000:
        tier = "Luxury Premium Product"
    elif price >= 5000:
        tier = "Premium Product"
    elif price >= 1000:
        tier = "Expensive Product"
    else:
        tier = "Affordable Product"
    print(f"{item:11s} (Rs.{price:6,}) -> {tier}")
```

### Task 7: Student Exam Eligibility (Attendance & Marks)
**Conditions:**
- `Attendance >= 75%`:
  - If `Marks >= 40`: Eligible for Semester Pass Certificate
  - Else: Failed in Exam (Eligible for Re-examination)
- `Attendance < 75%`: Not Eligible (Barred from Exam due to shortage of attendance)

```python
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit":  {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha":  {"attendance": 78, "marks": 55}
}

print("--- Task 7: Student Exam Eligibility ---")
for name, record in students.items():
    att = record["attendance"]
    marks = record["marks"]
    
    if att >= 75:
        if marks >= 40:
            status = "Eligible & Passed Examination"
        else:
            status = "Attended Exam but Failed (Eligible for Backlog / Re-exam)"
    else:
        status = "Not Eligible (Barred due to low attendance < 75%)"
        
    print(f"{name:6s} [Att: {att}%, Marks: {marks:2d}] -> {status}")
```

### Task 8: Bank Account Status Classification
**Conditions:**
- `Balance > 10,000`: Premium Account
- `Balance > 0`: Regular Account
- `Balance == 0`: Zero Balance Account
- `Balance < 0`: Overdraft / Penalty Account

```python
accounts = {
    "Rahul": 25000,
    "Amit":  5000,
    "Priya": 0,
    "Neha":  -2000
}

print("--- Task 8: Bank Account Status ---")
for customer, balance in accounts.items():
    if balance > 10000:
        account_type = "Premium Account (High Net Worth)"
    elif balance > 0:
        account_type = "Regular Active Account"
    elif balance == 0:
        account_type = "Zero Balance Account"
    else:
        account_type = "Overdraft Negative Balance (Action Required)"
    print(f"{customer:6s} : Rs.{balance:7,} -> {account_type}")
```

### Task 9: Nested Employee Performance & Compensation Review
Evaluates salary, experience, and performance rating for promotion and increments:

```python
staff = {
    "Rahul": {"salary": 65000, "experience": 6, "score": 92},
    "Amit":  {"salary": 42000, "experience": 2, "score": 88},
    "Priya": {"salary": 72000, "experience": 7, "score": 74},
    "Neha":  {"salary": 48000, "experience": 3, "score": 95}
}

print("--- Task 9: Employee Performance Review ---")
for name, data in staff.items():
    sal, exp, score = data["salary"], data["experience"], data["score"]
    
    if sal >= 50000:
        if score >= 90:
            review = "Senior Level: Outstanding (Double Increment & Promotion)"
        elif score >= 75:
            review = "Senior Level: Strong Performance (Standard Increment)"
        else:
            review = "Senior Level: Needs Improvement"
    else:
        if score >= 90:
            review = "Junior Level: Fast-Track Candidate (High Increment)"
        else:
            review = "Junior Level: Standard Progression"
            
    print(f"{name:6s} [Exp: {exp}y, Score: {score}, Sal: Rs.{sal:,}] -> {review}")
```

### Task 10: Comprehensive Attendance & Academic Division System
Combines attendance check with academic divisions:

```python
students = {
    "Rahul": {"attendance": 85, "marks": 82},
    "Amit":  {"attendance": 65, "marks": 90},
    "Priya": {"attendance": 90, "marks": 52},
    "Rohit": {"attendance": 80, "marks": 36}
}

print("--- Task 10: Academic Result Evaluation ---")
for name, rec in students.items():
    att = rec["attendance"]
    marks = rec["marks"]
    
    if att < 75:
        verdict = "Not Eligible for Result (Attendance Shortage < 75%)"
    else:
        if marks >= 80:
            verdict = "Passed: First Division with Distinction"
        elif marks >= 60:
            verdict = "Passed: First Division"
        elif marks >= 40:
            verdict = "Passed: Second Division"
        else:
            verdict = "Failed in Exam (Marks < 40)"
            
    print(f"{name:6s} [Att: {att}%, Marks: {marks:2d}] -> {verdict}")
```
