### Syntax Overview
- **Basic Mapping**: `[expression for item in iterable]`
- **With Filter (if)**: `[expression for item in iterable if condition]`
- **With Conditional Expression (if-else)**: `[expression_if_true if condition else expression_if_false for item in iterable]`

```python
# Basic mapping & filtering demo
data = [1, 2, 3, 4, 5]
squares = [x ** 2 for x in data]
evens = [x for x in data if x % 2 == 0]

print("Original data:", data)
print("Squares:", squares)
print("Evens:  ", evens)
```

### Q1: Find numbers greater than 50
**Given:** `numbers = [20, 55, 30, 80, 45, 90, 10]`
**Expected:** `[55, 80, 90]`

```python
numbers = [20, 55, 30, 80, 45, 90, 10]
result = [num for num in numbers if num > 50]
print("Numbers > 50:", result)
```

### Q2: Find numbers between 20 and 50
**Given:** `numbers = [10, 25, 35, 60, 45, 15, 50, 70]`
**Expected:** `[25, 35, 45, 50]`

```python
numbers = [10, 25, 35, 60, 45, 15, 50, 70]
result = [num for num in numbers if 20 <= num <= 50]
print("Numbers between 20 and 50:", result)
```

### Q3: Find numbers divisible by both 3 and 5
**Given:** `numbers = [10, 15, 20, 30, 45, 50, 60]`
**Expected:** `[15, 30, 45, 60]`

```python
numbers = [10, 15, 20, 30, 45, 50, 60]
result = [num for num in numbers if num % 3 == 0 and num % 5 == 0]
print("Numbers divisible by 3 and 5:", result)
```

### Q4: Find names having more than 4 characters
**Given:** `names = ["Amit", "Priya", "Rahul", "Soniya", "Sumit", "John"]`
**Expected:** `['Priya', 'Rahul', 'Soniya', 'Sumit']`

```python
names = ["Amit", "Priya", "Rahul", "Soniya", "Sumit", "John"]
result = [name for name in names if len(name) > 4]
print("Names with length > 4:", result)
```

### Q5: Find square of only even numbers from a list
**Given:** `numbers = [1, 2, 3, 4, 5, 6, 7, 8]`
**Expected:** `[4, 16, 36, 64]`

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
result = [x ** 2 for x in numbers if x % 2 == 0]
print("Squares of even numbers:", result)
```

### Q6: Pass / Fail grade classification
**Given:** `marks = [76, 45, 80, 34, 23, 85]`
**Expected:** `['Pass', 'Pass', 'Pass', 'Fail', 'Fail', 'Pass']`

```python
marks = [76, 45, 80, 34, 23, 85]
result = ["Pass" if m >= 40 else "Fail" for m in marks]
print("Marks: ", marks)
print("Result:", result)
```

### Q7: Convert temperatures from Celsius to Fahrenheit
**Formula:** `(C * 9/5) + 32`
**Given:** `celsius = [0, 10, 20, 30, 40]`

```python
celsius = [0, 10, 20, 30, 40]
fahrenheit = [(c * 9/5) + 32 for c in celsius]
print("Celsius:   ", celsius)
print("Fahrenheit:", fahrenheit)
```

### Q8: Replace every even number with 'Even' and every odd number with 'Odd'
**Given:** `numbers = [10, 15, 20, 25, 30]`
**Expected:** `['Even', 'Odd', 'Even', 'Odd', 'Even']`

```python
numbers = [10, 15, 20, 25, 30]
result = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print("Numbers:", numbers)
print("Labels: ", result)
```

### Q9: Replace negative numbers with 0
**Given:** `numbers = [10, -5, 20, -8, 30, -2]`
**Expected:** `[10, 0, 20, 0, 30, 0]`

```python
numbers = [10, -5, 20, -8, 30, -2]
result = [x if x > 0 else 0 for x in numbers]
print("Original:", numbers)
print("Result:  ", result)
```

### Q10: Tiered product discount (10% for >= ₹1000, otherwise 5%)
**Given:** `prices = [500, 1200, 800, 2000, 1500]`

```python
prices = [500, 1200, 800, 2000, 1500]
discounted_prices = [p * 0.90 if p >= 1000 else p * 0.95 for p in prices]
print("Original prices:   ", prices)
print("Discounted prices: ", discounted_prices)
```

### Q11: Filter and clean names (Strip whitespaces & uppercase)
**Given:** `raw_names = [" Amit ", " Priya", "Rahul ", "Soniya"]`

```python
raw_names = [" Amit ", " Priya", "Rahul ", "Soniya"]
cleaned_upper = [name.strip().upper() for name in raw_names]
print("Cleaned & Uppercased:", cleaned_upper)
```

### Q12: Flatten a 2D matrix into a 1D list
**Given:** `matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]`

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flattened = [val for row in matrix for val in row]
print("2D Matrix:", matrix)
print("Flattened list:", flattened)
```

### Q13: E-commerce: Find products costing more than ₹1000
**Given:** `prices = [499, 1299, 799, 2499, 999, 1599]`
**Expected:** `[1299, 2499, 1599]`

```python
prices = [499, 1299, 799, 2499, 999, 1599]
costly_products = [p for p in prices if p > 1000]
print("Products > Rs. 1000:", costly_products)
```

### Q14: E-commerce Discount: 20% discount on products > ₹2000
**Given:** `prices = [500, 1200, 2500, 3000, 1500, 5000]`
**Expected:** `[2000.0, 2400.0, 4000.0]`

```python
prices = [500, 1200, 2500, 3000, 1500, 5000]
discounted_over_2000 = [p * 0.80 for p in prices if p > 2000]
print("20% discounted prices for products > Rs. 2000:", discounted_over_2000)
```

### Q15: Bank Transactions: Filter deposits and withdrawals
**Given:** `transactions = [5000, -2000, 3000, -1000, 7000, -500]`

```python
transactions = [5000, -2000, 3000, -1000, 7000, -500]

deposits = [tx for tx in transactions if tx > 0]
withdrawals = [tx for tx in transactions if tx < 0]

print("All transactions: ", transactions)
print("Deposits (+):     ", deposits)
print("Withdrawals (-):  ", withdrawals)
```

### Q16: Student Database: Extract names of students scoring >= 60
**Given:** Student records list of dictionaries

```python
students = [
    {"name": "Amit",  "marks": 75},
    {"name": "Rahul", "marks": 90},
    {"name": "Neha",  "marks": 55},
    {"name": "Rohit", "marks": 68}
]

top_students = [s["name"] for s in students if s["marks"] >= 60]
print("Students scoring 60 or more:", top_students)
```

### Q17: Employee Database: Extract names of IT department employees
**Given:** Employee records with salary and department

```python
employees = [
    {"name": "Amit",  "department": "HR",      "salary": 45000},
    {"name": "Rahul", "department": "IT",      "salary": 60000},
    {"name": "Neha",  "department": "IT",      "salary": 75000},
    {"name": "Rohit", "department": "Finance", "salary": 35000}
]

it_employees = [e["name"] for e in employees if e["department"] == "IT"]
high_earning_employees = [e["name"] for e in employees if e["salary"] > 50000]

print("Employees in IT Department:", it_employees)
print("Employees with salary > 50k:", high_earning_employees)
```
