### Exercise 8: Take a number and print its multiplication table

```python
def print_multiplication_table(num):
    print(f"Multiplication Table for {num}:")
    for i in range(1, 11):
        print(f"{num} x {i:2d} = {num * i}")

# Demo with number 12
print_multiplication_table(12)
```

### Exercise 10: Count even and odd numbers between 1 and 50

```python
count_even = 0
count_odd = 0

for i in range(1, 51):
    if i % 2 == 0:
        count_even += 1
    else:
        count_odd += 1

print(f"Total even numbers (1 to 50): {count_even}")
print(f"Total odd numbers (1 to 50): {count_odd}")
```

### Exercise 11: Print the elements of a list in reverse order
List: `[22, 33, 12, 32, 45, 13]`

```python
data = [22, 33, 12, 32, 45, 13]
print("Original list:", data)

print("Reversed order elements:")
for i in range(len(data) - 1, -1, -1):
    print(data[i], end=" ")
print()
```

### Exercise 12: Find the largest number in a list using a `for` loop
List: `[22, 55, 12, 89, 34, 67]`

```python
data = [22, 55, 12, 89, 34, 67]

largest = data[0]
for num in data:
    if num > largest:
        largest = num

print("List:", data)
print("Largest number in list:", largest)
```

### Exercise 13: Find the smallest number in a list using a `for` loop
List: `[22, 55, 12, 89, 34, 67]`

```python
data = [22, 55, 12, 89, 34, 67]

smallest = data[0]
for num in data:
    if num < smallest:
        smallest = num

print("List:", data)
print("Smallest number in list:", smallest)
```

### Exercise 14: Count vowels in a string

```python
sentence = "Python Programming Language"
vowels = "aeiouAEIOU"
vowel_count = 0

for char in sentence:
    if char in vowels:
        vowel_count += 1

print(f"Sentence: '{sentence}'")
print(f"Total vowels count: {vowel_count}")
```

### Exercise 15: Count the number of spaces in a sentence
Sentence: `"Python is easy to learn"`

```python
text = "Python is easy to learn"
space_count = 0

for char in text:
    if char == " ":
        space_count += 1

print(f"Text: '{text}'")
print(f"Number of space characters: {space_count}")
```

### Exercise 16: Count positive and negative numbers using a `while` loop
List: `[22, 32, 45, -2, 87, -1, 89, -4]`

```python
data = [22, 32, 45, -2, 87, -1, 89, -4]

pos_count = 0
neg_count = 0
idx = 0

while idx < len(data):
    if data[idx] > 0:
        pos_count += 1
    elif data[idx] < 0:
        neg_count += 1
    idx += 1

print("Dataset:", data)
print("Positive numbers count:", pos_count)
print("Negative numbers count:", neg_count)
```

### Exercise 17: Count occurrences of a character using a `while` loop
String: `"Python Programming p"`

```python
data = "Python Programming p"
target_char = "p"
count = 0
idx = 0

# Case-insensitive search demonstration
while idx < len(data):
    if data[idx].lower() == target_char.lower():
        count += 1
    idx += 1

print(f"String: '{data}'")
print(f"Occurrences of '{target_char}' (case-insensitive): {count}")
```
