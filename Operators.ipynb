# Python Operators

This notebook covers all fundamental operators in Python as presented in the coursework images:
- Arithmetic Operators
- Compound Assignment Operators
- Relational / Comparison Operators
- Logical Operators (`and`, `or`, `not`)
- Identity Operators (`is`, `is not` vs `==`)
- Membership Operators (`in`, `not in`)
- Unary Operations and Increment/Decrement semantics in Python

## 1. Arithmetic Operators
Used to perform mathematical operations:
- `+` Addition / String Concatenation
- `-` Subtraction
- `*` Multiplication / Sequence Repetition
- `/` True Division (returns float)
- `//` Floor Division (rounds down to integer)
- `%` Modulus (remainder)
- `**` Exponentiation (power)

```python
x = 20
y = 6

print("Addition (x + y):", x + y)
print("Subtraction (x - y):", x - y)
print("Multiplication (x * y):", x * y)
print("True Division (x / y):", x / y)
print("Floor Division (x // y):", x // y)
print("Modulus (x % y):", x % y)
print("Power (x ** 2):", x ** 2)

# String operations with + and *
firstName = "Amit"
lastName = "Kumar"
print("Full Name:", firstName + " " + lastName)
print("String Repetition:", "Python! " * 3)
```

## 2. Compound Assignment Operators
Used to perform an operation and assign the result back to the same variable in-place:
`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`

```python
num = 10
print("Initial value:", num)

num += 5   # num = num + 5
print("After num += 5:", num)

num -= 3   # num = num - 3
print("After num -= 3:", num)

num *= 2   # num = num * 2
print("After num *= 2:", num)

num /= 4   # num = num / 4
print("After num /= 4:", num)

num //= 2  # num = num // 2
print("After num //= 2:", num)

num %= 2   # num = num % 2
print("After num %= 2:", num)
```

## 3. Relational / Comparison Operators
Used to compare two values. The result is always a Boolean (`True` or `False`):
- `==` Equal to
- `!=` Not equal to
- `>` Greater than
- `<` Less than
- `>=` Greater than or equal to
- `<=` Less than or equal to

```python
a = 15
b = 25

print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= 15:", a >= 15)
print("b <= 20:", b <= 20)
```

## 4. Logical Operators
Used to combine conditional statements:
- `and`: Returns `True` only if BOTH conditions are `True`
- `or`: Returns `True` if AT LEAST ONE condition is `True`
- `not`: Inverts the Boolean value (`not True` -> `False`)

```python
x = 10
y = 20

# Logical AND
print("x < y and y > 15:", x < y and y > 15)      # True and True -> True
print("x > y and y > 15:", x > y and y > 15)      # False and True -> False
print("x < y and y < 15:", x < y and y < 15)      # True and False -> False
print("x > y and y < 15:", x > y and y < 15)      # False and False -> False

# Logical OR
print("\nx > y or y > 15:", x > y or y > 15)       # False or True -> True
print("x > y or y < 15:", x > y or y < 15)        # False or False -> False

# Logical NOT
print("\nnot (x < y):", not (x < y))               # not True -> False
print("not (x > y):", not (x > y))               # not False -> True
```

## 5. Identity Operators (`is` and `is not`) vs Equality (`==`)
- `==` checks whether the **values/contents** of two objects are equal.
- `is` checks whether two variables refer to the **exact same object in memory** (same memory address / `id()`).
- `is not` returns `True` if both variables point to different objects.

```python
# Case 1: Variables referencing the same list object
x = [1, 2, 3]
y = x  # y points to the exact same list object in memory
print("x is y:", x is y)            # True: Same memory reference
print("x == y:", x == y)            # True: Same values
print("id(x) == id(y):", id(x) == id(y))

# Case 2: Variables having equal values but different objects in memory
a = [1, 2, 3]
b = [1, 2, 3]
print("\na is b:", a is b)            # False: Different objects in memory (mutable lists)
print("a == b:", a == b)            # True: Values are identical
print("a is not b:", a is not b)    # True
print("id(a):", id(a), "vs id(b):", id(b))

# Note on small integers and interned strings
num1 = 100
num2 = 100
print("\nnum1 is num2 (interned small ints):", num1 is num2)
```

## 6. Membership Operators (`in` and `not in`)
Used to check whether a specific value exists inside a sequence or collection (such as list, tuple, string, set, dictionary):

```python
data = [10, 20, 66, 88]

# Checking membership in list
print("20 in data:", 20 in data)
print("99 in data:", 99 in data)
print("99 not in data:", 99 not in data)

# Checking membership in string
course = "Python Programming"
print("'Python' in course:", "Python" in course)
print("'Java' in course:", "Java" in course)

# Checking membership in dictionary (checks keys by default)
student = {"name": "Mohit", "age": 23, "active": True}
print("'name' in student:", "name" in student)
print("'college' in student:", "college" in student)
print("'Mohit' in student.values():", "Mohit" in student.values())
```

## 7. Python Increment / Decrement Semantics
**Important Rule**: Python does **NOT** support `++` (post/pre-increment) or `--` (post/pre-decrement) operators like C/C++/Java.
- In Python, `++i` is evaluated as `+(+i)` (unary plus), which does NOT change `i`.
- In Python, `--i` is evaluated as `-(-i)` (unary negation twice), which does NOT change `i`.
- To increment or decrement in Python, always use `i += 1` or `i -= 1`.

```python
i = 5
print("Initial i:", i)

# Evaluating ++i (unary plus, not increment!)
result = ++i
print("Value of ++i:", result, "| Value of i after ++i:", i)

# Evaluating --i (unary minus minus, not decrement!)
result = --i
print("Value of --i:", result, "| Value of i after --i:", i)

# The correct Pythonic way:
i += 1
print("After i += 1 (correct increment):", i)

i -= 1
print("After i -= 1 (correct decrement):", i)
```
