### 1. Solid Rectangle Pattern (rows = 4, cols = 6)

```python
rows = 4
cols = 6

print(f"Rectangle Pattern ({rows}x{cols}):")
for i in range(rows):
    for j in range(cols):
        print("*", end=" ")
    print()
```

### 2. Right-Angled Star Triangle (`for` loop)

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

### 3. Right-Angled Number Triangle (1 to i)

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

### 4. Repeated Number Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i, end=" ")
    print()
```

### 5. Floyd's Continuous Number Triangle (1 to 15)

```python
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(f"{num:2d}", end=" ")
        num += 1
    print()
```

### 6. Binary Alternating Triangle (1 and 0)

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        # Even sum gives 1, Odd sum gives 0
        if (i + j) % 2 == 0:
            print("1", end=" ")
        else:
            print("0", end=" ")
    print()
```

### 7. Alphabet Triangle using `chr(65+j)` (A, AB, ABC...)

```python
n = 5
for i in range(1, n + 1):
    for j in range(0, i):
        print(chr(65 + j), end=" ")
    print()
```

### 8. Repeated Alphabet Triangle (A, BB, CCC...)

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(chr(64 + i), end=" ")
    print()
```

### 9. Inverted Right-Angled Star Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

### 10. Inverted Number Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

### 11. Pyramid Star Pattern (Centered with Spaces)

```python
n = 5
for i in range(1, n + 1):
    # Print leading spaces
    for s in range(n - i):
        print(" ", end=" ")
    # Print stars (2*i - 1)
    for st in range(2 * i - 1):
        print("*", end=" ")
    print()
```

### 12. Full Diamond Star Pattern (9 Rows total)

```python
n = 5  # Half height

# Upper half including middle row (1 to 5)
for i in range(1, n + 1):
    for s in range(n - i):
        print(" ", end=" ")
    for st in range(2 * i - 1):
        print("*", end=" ")
    print()

# Lower inverted half (4 down to 1)
for i in range(n - 1, 0, -1):
    for s in range(n - i):
        print(" ", end=" ")
    for st in range(2 * i - 1):
        print("*", end=" ")
    print()
```

### 13. Palindromic Number Pyramid

```python
n = 5
for i in range(1, n + 1):
    # Spaces
    for s in range(n - i):
        print(" ", end=" ")
    # Increasing numbers (1 to i)
    for j in range(1, i + 1):
        print(j, end=" ")
    # Decreasing numbers (i-1 down to 1)
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()
```

### 14. Right-Angled Star Triangle using `while` loop

```python
num = 5
row = 1

while row <= num:
    star = 1
    while star <= row:
        print("*", end=" ")
        star += 1
    print()
    row += 1
```

### 15. Inverted Star Triangle using `while` loop

```python
num = 5
row = 1

while row <= num:
    star = 1
    while star <= (num - row + 1):
        print("*", end=" ")
        star += 1
    print()
    row += 1
```

### 16. Mirrored Right-Angled Triangle using `while` loop

```python
num = 5
row = 1

while row <= num:
    # Print spaces
    space = 1
    while space <= (num - row):
        print(" ", end=" ")
        space += 1
    # Print stars
    star = 1
    while star <= row:
        print("*", end=" ")
        star += 1
    print()
    row += 1
```
