### 2. Basic QR Code Generation from Text / URL
Using `qrcode.make(data)` to generate and save a basic QR code:

### 3. URL Validation and QR Code Generator
**Condition:** Ensure the URL starts with `http://` or `https://` before creating the QR code.

````python
def generate_validated_url_qr(url, output_filename="Validated_URL_QR.png"):
    print(f"Validating URL: '{url}'")
    
    if url.startswith("http://") or url.startswith("https://"):
        qr = qrcode.make(url)
        qr.save(output_filename)
        print(f"Success: QR Code Ban Gaya! Saved to '{output_filename}'.")
    else:
        print("Error: Invalid URL! URL must start with 'http://' or 'https://'.")

# Valid URL test
generate_validated_url_qr("https://github.com/nitin")

# Invalid URL test
generate_validated_url_qr("ftp://invalid-protocol.com")
````

### 4. Customized QR Code Styling
Customizing version, error correction level (`ERROR_CORRECT_H`), box size, border, fill color, and background color:

````python
import qrcode
from qrcode.constants import ERROR_CORRECT_H

def generate_custom_qr(data, fill_color="blue", back_color="white", filename="Custom_Styled_QR.png"):
    qr = qrcode.QRCode(
        version=1,
        error_correction=ERROR_CORRECT_H,
        box_size=10,
        border=4
    )
    qr.add_data(data)
    qr.make(fit=True)
    
    img = qr.make_image(fill_color=fill_color, back_color=back_color)
    img.save(filename)
    print(f"Custom styled QR ({fill_color} on {back_color}) saved as '{filename}'!")

generate_custom_qr("https://www.python.org", fill_color="darkblue", back_color="lightyellow")
````

### 5. Dynamic UPI Payment QR Code Generator
Generates a standard UPI deep-link QR code:
`upi://pay?pa={upi_id}&pn={payee_name}`

````python
def create_upi_qr(upi_id, payee_name, filename="UPI_QR.png"):
    # Constructing standard UPI URL
    upi_url = f"upi://pay?pa={upi_id}&pn={payee_name}"
    
    print(f"Generating UPI QR for Payee: {payee_name} ({upi_id})")
    print(f"Payload URL: {upi_url}")
    
    qr = qrcode.make(upi_url)
    qr.save(filename)
    print(f"UPI Ka QR Ban Gaya..!! Saved to '{filename}'\n")

create_upi_qr("akhilesh@okaxis", "Akhilesh Kumar", "Akhilesh_UPI.png")
````

### 6. Dynamic UPI QR Code with Custom Amount & Currency
Generates a UPI deep-link with pre-set payment amount:
`upi://pay?pa={upi_id}&pn={payee_name}&am={amount}&cu=INR`

````python
def create_upi_payment_qr(upi_id, payee_name, amount, filename="UPI_Invoice_QR.png"):
    upi_url = f"upi://pay?pa={upi_id}&pn={payee_name}&am={amount:.2f}&cu=INR"
    
    print(f"--- UPI Invoice Generation ---")
    print(f"Payee Name: {payee_name}")
    print(f"UPI ID:     {upi_id}")
    print(f"Amount:     Rs. {amount:.2f}")
    print(f"Payload:    {upi_url}")
    
    qr = qrcode.make(upi_url)
    qr.save(filename)
    print(f"UPI Amount QR Code Ban Gaya! Saved to '{filename}'\n")

create_upi_payment_qr("akhilesh@okaxis", "Akhilesh Kumar", 1499.00, "Akhilesh_Invoice_1499.png")
````

## Part B: Optical Character Recognition (OCR)

### 7. Extracting Text from Images with `pytesseract` and `PIL`
`pytesseract` is a Python wrapper for Google's Tesseract-OCR Engine:

````python
import os
from PIL import Image

try:
    import pytesseract
    
    # Common standard installation paths on Windows (using forward slashes)
    common_tesseract_paths = [
        "C:/Program Files/Tesseract-OCR/tesseract.exe",
        "C:/Program Files (x86)/Tesseract-OCR/tesseract.exe",
        os.path.expanduser("~/AppData/Local/Programs/Tesseract-OCR/tesseract.exe")
    ]
    for p in common_tesseract_paths:
        if os.path.exists(p):
            pytesseract.pytesseract.tesseract_cmd = p
            print(f"Configured Tesseract binary at: {p}")
            break
            
    print("pytesseract module is ready for OCR processing.")
except ImportError:
    print("pytesseract is not installed. Install with: pip install pytesseract")
````

### 8. OCR Extraction Function with Fallback Handling

````python
def extract_text_from_image(image_path):
    print(f"Performing OCR on: '{image_path}'")
    
    if not os.path.exists(image_path):
        print(f"Error: Image file '{image_path}' not found.")
        return ""
        
    try:
        import pytesseract
        img = Image.open(image_path)
        extracted_text = pytesseract.image_to_string(img)
        print("--- Extracted Text ---")
        print(extracted_text if extracted_text.strip() else "[No text detected in image]")
        return extracted_text
    except Exception as e:
        print("OCR Processing Note:", e)
        print("Tip: Ensure Tesseract OCR engine is installed on your system.")
        return ""

# Demo with one of the generated QR codes
extract_text_from_image("Basic_QR.png")
````
