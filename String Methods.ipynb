## 1. String Creation, Length, and Immutability
Strings in Python are immutable sequences of characters. Once created, individual characters cannot be modified in-place.

```python
data = "Python Programming"
print("String:", data)
print("Type:", type(data))
print("Length:", len(data))

# Immutability Demonstration:
try:
    data[0] = "J"  # Raises TypeError
except TypeError as e:
    print("\nImmutability check - TypeError caught:", e)
```

## 2. Indexing and Slicing
- **Positive indexing**: `0` to `len - 1`
- **Negative indexing**: `-1` (last char) to `-len` (first char)
- **Slicing syntax**: `string[start : stop : step]`

```python
course = "python"

# Indexing
print("First character course[0]:", course[0])
print("Last character course[-1]:", course[-1])
print("Character at index 2:", course[2])

# Slicing
print("\nSlicing course[0:6]:", course[0:6])
print("Slicing course[0:2]:", course[0:2])
print("Slicing course[2:]:", course[2:])
print("Slicing course[:4]:", course[:4])
print("Reverse using slicing course[::-1]:", course[::-1])
print("Every 2nd char course[::2]:", course[::2])
```

## 3. String Repetition (`*`) and Concatenation (`+`)

```python
data = "Python "
print("Repetition (data * 3):", data * 3)

greeting = "Hello"
target = "World"
full_message = greeting + ", " + target + "!"
print("Concatenation:", full_message)
```

## 4. String Traversal & Character Counting with Loops

```python
text = "Python Programming p"

# Traversal with for loop
print("Characters in text:")
for ch in "PYTHON":
    print(ch, end=" - ")
print("\n")

# Count occurrences of 'p' or 'P' using while loop
target_char = 'p'
count = 0
i = 0
while i < len(text):
    if text[i].lower() == target_char.lower():
        count += 1
    i += 1

print(f"Total count of '{target_char}' in '{text}': {count}")
```

## 5. Case Conversion Methods
- `upper()`: Converts all characters to uppercase
- `lower()`: Converts all characters to lowercase
- `title()`: Capitalizes first letter of each word
- `capitalize()`: Capitalizes only first letter of string
- `swapcase()`: Inverts case of each character

```python
text = "python programming language"

print("Original:   ", text)
print("upper():    ", text.upper())
print("lower():    ", text.lower())
print("title():    ", text.title())
print("capitalize():", text.capitalize())
print("swapcase():  ", text.title().swapcase())
```

## 6. Whitespace Stripping Methods
- `strip()`: Removes leading and trailing whitespaces
- `lstrip()`: Removes only leading (left) whitespaces
- `rstrip()`: Removes only trailing (right) whitespaces

```python
padded_text = "   Python Programming   "

print(f"Raw string:     '{padded_text}'")
print(f"strip():        '{padded_text.strip()}'")
print(f"lstrip():       '{padded_text.lstrip()}'")
print(f"rstrip():       '{padded_text.rstrip()}'")
```

## 7. Substring Search: `find()` vs `index()`
- `find(sub)`: Returns the lowest index where substring is found; returns `-1` if not found (never crashes).
- `index(sub)`: Like `find()`, but raises `ValueError` if substring is not found.
- `count(sub)`: Returns number of non-overlapping occurrences.

```python
data = "Python Programming"

# find() demonstration
print("data.find('P'):", data.find('P'))
print("data.find('Programming'):", data.find('Programming'))
print("data.find('z'):", data.find('z'))  # Returns -1

# index() demonstration
print("\ndata.index('P'):", data.index('P'))
try:
    print(data.index('z'))
except ValueError as e:
    print("data.index('z') raised ValueError:", e)

# count() demonstration
print("\ndata.count('m'):", data.count('m'))
print("data.count('P'):", data.count('P'))
```

## 8. Prefix and Suffix Checks: `startswith()` & `endswith()`
Commonly used for validating email domains, URLs, and file extensions:

```python
email = "admin@gmail.com"

print("Email:", email)
print("email.startswith('admin'):", email.startswith("admin"))
print("email.startswith('user'):", email.startswith("user"))
print("email.endswith('.com'):", email.endswith(".com"))
print("email.endswith('.org'):", email.endswith(".org"))
print("'@' in email:", "@" in email)
```

## 9. Character Validation Methods
- `isalpha()`: Returns `True` if all characters are alphabetic letters
- `isdigit()`: Returns `True` if all characters are digits (0-9)
- `isalnum()`: Returns `True` if all characters are alphanumeric (letters or numbers)
- `isspace()`: Returns `True` if string contains only whitespace characters

```python
s1 = "Invertis"
s2 = "12345"
s3 = "Invertis12345"
s4 = "Invertis@12345"
s5 = "   "

print(f"'{s1}'.isalpha():", s1.isalpha())
print(f"'{s2}'.isdigit():", s2.isdigit())
print(f"'{s3}'.isalnum():", s3.isalnum())
print(f"'{s4}'.isalnum():", s4.isalnum())  # False due to '@'
print(f"'{s5}'.isspace():", s5.isspace())
```

## 10. Splitting and Joining: `split()` and `join()`

```python
sentence = "I Love Python Programming"

# Splitting string into list of words
words = sentence.split(" ")
print("Original Sentence:", sentence)
print("words list after split(' '):", words)

# Joining list elements back with a delimiter
rejoined_with_hyphen = "-".join(words)
print("Rejoined with '-':", rejoined_with_hyphen)

csv_data = "Amit,MCA,23,80000"
fields = csv_data.split(",")
print("\nCSV parsed fields:", fields)
```

## 11. String Replacement: `replace()`
`string.replace(old, new, count)` replaces occurrences of `old` with `new`.

```python
message = "Java is great. Java is fast."
new_message = message.replace("Java", "Python")
print("Original:", message)
print("Replaced:", new_message)

# Replace only first occurrence
single_replace = message.replace("Java", "Python", 1)
print("First occurrence only:", single_replace)
```

## 12. String Formatting: f-strings and `.format()`

```python
name = "Sumit"
profile = "HR"

# Method 1: Using commas in print
print("1. Comma separation: User name is:", name, "and profile is:", profile)

# Method 2: Using .format()
print("2. str.format(): User name is: {} and his profile is: {}".format(name, profile))
print("   Named arguments: User name is: {n} and role is: {p}".format(n=name, p=profile))

# Method 3: f-strings (Modern Python 3.6+)
print(f"3. f-string: User name is: {name} and his profile is: {profile}")
salary = 65000.758
print(f"   Formatted currency: Rs. {salary:,.2f}")
```
