## Part A: Tuples

### 1. Tuple Creation (The Comma Rule)
In Python, parentheses `()` are optional. It is the comma `,` that actually defines a tuple:

```python
# Creation with parentheses
t1 = (10, 20, 30)
print("t1:", t1, "| type:", type(t1))

# Creation without parentheses (comma makes it a tuple)
t2 = 10, 20, 30
print("t2:", t2, "| type:", type(t2))
```

### 2. Single-Element Tuple Syntax
**Important Rule**: A single item in parentheses without a comma is treated as a regular integer/string/float. To create a 1-element tuple, you MUST include a trailing comma:

```python
item1 = (10)
print("item1 = (10)      -> type:", type(item1))  # <class 'int'>

item2 = (10,)
print("item2 = (10,)     -> type:", type(item2))  # <class 'tuple'>

item3 = "Amit",
print("item3 = 'Amit',   -> type:", type(item3))  # <class 'tuple'>
```

### 3. Immutability Demonstration & TypeError
Tuples are immutable; their elements cannot be changed, added, or deleted after creation:

```python
data = (10, 20, 30, 40)
print("Tuple data:", data)

try:
    data[0] = 88  # Attempting item assignment
except TypeError as e:
    print("TypeError successfully caught:", e)
```

### 4. Tuple Indexing and Slicing

```python
t = (10, 20, 30, 40, 50)

print("Positive index t[2]:", t[2])
print("Negative index t[-1]:", t[-1])
print("Slicing t[1:4]:", t[1:4])
print("Reverse t[::-1]:", t[::-1])
```

### 5. Tuple Packing and Unpacking

```python
# Tuple Packing: Multiple values packed into a single tuple
record = "Amit", "MCA", 23, True
print("Packed record:", record)

# Tuple Unpacking: Extracting elements into individual variables
name, course, age, enrolled = record
print(f"Unpacked -> Name: {name}, Course: {course}, Age: {age}, Enrolled: {enrolled}")
```

### 6. Extended Unpacking with `*` (Star syntax)
Allows one variable to capture multiple elements as a list:

```python
record = ("Amit", "MCA", 23, 80000.00, "Bareilly", "India")

name, course, *other_details = record
print("Name:", name)
print("Course:", course)
print("Other details collected by *:", other_details)
```

### 7. Unpacking Mismatch ValueError
If the number of variables does not match the tuple elements (without `*`), Python raises a `ValueError`:

```python
record = ("Amit", 23, True)

try:
    name, age = record  # 3 elements but only 2 variables
except ValueError as e:
    print("ValueError caught:", e)
```

### 8. Tuple Methods: `count()` and `index()`
Tuples only have two built-in methods because they are immutable:

```python
numbers = (10, 20, 30, 20, 40, 20, 50)

print("Numbers tuple:", numbers)
print("numbers.count(20):", numbers.count(20))
print("numbers.index(30):", numbers.index(30))
print("numbers.index(20):", numbers.index(20))  # First occurrence
```

### 9. Tuple Built-in Functions

```python
data = (10, 50, 30, 90, 20)

print("len():", len(data))
print("min():", min(data))
print("max():", max(data))
print("sum():", sum(data))
```

## Part B: Sets

### 10. Set Creation (`{}` vs `set()`)
- `{}` creates an empty **dictionary**.
- `set()` creates an empty **set**.

```python
empty_dict = {}
print("type of {}:", type(empty_dict))

empty_set = set()
print("type of set():", type(empty_set))

# Initializing set with values
num_set = {10, 20, 30, 40}
print("num_set:", num_set, "| type:", type(num_set))
```

### 11. Automatic Duplicate Elimination
Sets only store unique elements:

```python
raw_list = [56, 22, 11, 10, 43, 76, 22, 11, 56]
unique_set = set(raw_list)

print("Original list with duplicates:", raw_list)
print("Set after deduplication:      ", unique_set)
```

### 12. Sets are Unordered & Unsubscriptable
Sets do not maintain index positions; accessing elements by index `[0]` raises `TypeError`:

```python
s = {10, 20, 30}

try:
    print(s[0])
except TypeError as e:
    print("TypeError caught when indexing a set:", e)
```

### 13. Set Modification Methods

```python
data_set = {10, 20, 30}
print("Initial set:", data_set)

# add() a single element
data_set.add(40)
print("After add(40):", data_set)

# update() with multiple elements
data_set.update([50, 60, 20])
print("After update([50, 60, 20]):", data_set)

# remove() vs discard()
data_set.remove(60)  # Removes 60
print("After remove(60):", data_set)

# discard() does NOT raise error if element not found
data_set.discard(999)
print("After discard(999) (no error):", data_set)

# pop() removes an arbitrary element
popped_elem = data_set.pop()
print(f"Popped '{popped_elem}':", data_set)

# clear()
data_set.clear()
print("After clear():", data_set)
```

### 14. Mathematical Set Operations (Union, Intersection, Difference)

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Set A:", A)
print("Set B:", B)

# Union (|)
print("\nUnion (A | B):", A.union(B))

# Intersection (&)
print("Intersection (A & B):", A.intersection(B))

# Difference (-)
print("Difference (A - B):", A.difference(B))
print("Difference (B - A):", B.difference(A))

# Symmetric Difference (^)
print("Symmetric Difference (A ^ B):", A.symmetric_difference(B))
```
